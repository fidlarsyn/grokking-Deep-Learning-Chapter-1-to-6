# **What is Deep Learning?**

*Deep learning* adalah subhimpunan (*subset*) metode dari *machine learning* yang berfokus pada penggunaan jaringan saraf tiruan (*artificial neural networks*). Algoritma ini terinspirasi dari struktur otak manusia untuk menyelesaikan tugas-tugas praktis industri seperti *computer vision* (pengolahan gambar), *Natural Language Processing* (pengolahan teks), dan pengenalan suara. *Deep learning* tidak hanya mengejar kecerdasan buatan umum (*General AI*), melainkan menyelesaikan masalah otomatisasi spesifik di industri.

# **What is Machine Learning?**

Definisi umum *machine learning* adalah subbidang ilmu komputer di mana komputer mempelajari cara menyelesaikan tugas tanpa diprogram secara eksplisit (*machines learn without being explicitly programmed*). Prinsip dasarnya adalah *"monkey see, monkey do"*: mesin mengamati pola dalam data dan berusaha meniru pola tersebut secara langsung (*supervised*) maupun tidak langsung (*unsupervised*)

# **Supervised Machine Learning**

*Supervised learning* adalah metode pembelajaran yang mengubah (*transform*) satu dataset input (*what you know*) menjadi dataset output (*what you want to know*). Contohnya termasuk mengubah piksel gambar menjadi label (ada kucing/tidak), mengubah riwayat film menjadi rekomendasi, atau mengubah data sensor cuaca menjadi probabilitas hujan. Pembelajaran dilakukan dengan meniru pemetaan langsung antara kedua dataset tersebut.

# **Unsupervised Machine Learning**

*Unsupervised learning* juga mengonversi satu dataset ke dataset lain, namun hasil akhirnya tidak diketahui atau tidak diberi label sebelumnya (tidak ada jawaban benar / *no target label*). Contoh utamanya adalah *clustering* (pengelompokan data), di mana algoritma mengelompokkan data ke dalam label kelompok (misalnya kelompok 1, 2, dst.) berdasarkan kemiripan struktur data tanpa diberi tahu nama kelompoknya.

# **Parametric vs. Nonparametric Learning**

* **Parametric Model**: Memiliki jumlah parameter (knob/bobot) yang bernilai tetap (*fixed number of parameters*). Proses belajarnya berorientasi pada pendekatan *trial-and-error* (mencari atau memutar knob hingga kesalahan berkurang).
* **Nonparametric Model**: Jumlah parameternya tidak terbatas dan ditentukan secara dinamis oleh ukuran data (*determined by data*). Metode ini umumnya berbasis perhitungan frekuensi atau probabilitas kemunculan (*counting and probability*).

# **Supervised Parametric Learning**

Mengombinasikan pendekatan *supervised* (pemetaan input ke target) dengan model *parametric* (menggunakan knob/bobot tetap). Proses pembelajaran terdiri dari 3 langkah utama:
1. **Predict**: Mengalirkan data input melalui konfigurasi knob/bobot saat ini untuk menghasilkan estimasi/prediksi.
2. **Compare**: Membandingkan hasil prediksi dengan nilai sebenarnya (*truth pattern*) untuk mengukur seberapa besar kesalahan (*error*).
3. **Learn**: Memutar atau menggeser sudut knob (penyesuaian bobot) agar prediksi berikutnya lebih akurat pada data serupa.

# **Unsupervised Parametric Learning**

Menggunakan knob/bobot untuk mengelompokkan data tanpa target label. Setiap kelompok memiliki konfigurasi knob yang memetakan probabilitas atau afinitas data input terhadap kelompok tersebut (misalnya menghasilkan nilai antara 0 hingga 1 sebagai tingkat keanggotaan kelompok).

# **Nonparametric Learning**

Algoritma yang menambahkan parameter baru saat menemukan hal baru untuk dihitung dalam data. Sebagai contoh, model menghitung jumlah kemunculan kondisi warna lampu lalu lintas yang menyebabkan mobil berjalan. Jika ada 3 warna lampu, terdapat 3 parameter hitungan; jika ada 5 warna lampu, terdapat 5 parameter.
