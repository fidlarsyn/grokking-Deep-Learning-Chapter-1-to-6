# **The Streetlight Problem &amp; Preparing the Data**

Memperkenalkan studi kasus nyata berupa **Streetlight Problem**. Jaringan saraf tiruan diajarkan untuk memprediksi apakah aman untuk berjalan (*walk = 1*) atau harus berhenti (*stop = 0*) berdasarkan status 3 lampu lalu lintas. Data diwakili dalam bentuk matriks di mana baris menyatakan sampel observasi dan kolom menyatakan kondisi masukan.

# **Matrices and Building a Neural Network**

Data masukan disusun ke dalam bentuk matriks 2D. Pembentukan jaringan saraf diawali dari arsitektur sederhana 3 masukan dan 1 keluaran. Operasi *forward propagation* dihitung menggunakan perkalian titik (*dot product*) antara vektor baris sampel masukan dengan vektor bobot.

```python
import numpy as np

weights = np.array([0.5, 0.48, -0.7])
alpha = 0.1

streetlights = np.array([
    [18]
])

walk_vs_stop = np.array([18])

input = streetlights
goal_prediction = walk_vs_stop

for iteration in range(20):
    prediction = input.dot(weights)
    error = (prediction - goal_prediction) ** 2
    delta = prediction - goal_prediction
    weights = weights - (alpha * (input * delta))
    print("Error:" + str(error) + " Prediction:" + str(prediction))
```

# **Learning the Whole Dataset &amp; Full, Batch, Stochastic Gradient Descent**

Untuk mempelajari seluruh dataset, jaringan dilatih melalui perulangan (*loop*) melewati semua baris sampel secara berurutan.
* **Stochastic Gradient Descent (SGD)**: Memperbarui bobot secara langsung setelah memproses **1 sampel data**.
* **Batch Gradient Descent**: Memperbarui bobot setelah menghitung rata-rata eror dari **sebagian sampel** (*mini-batch*).
* **Full Gradient Descent**: Memperbarui bobot hanya setelah menghitung total eror dari **seluruh dataset**.

```python
import numpy as np

weights = np.array([0.5, 0.48, -0.7])
alpha = 0.1

streetlights = np.array([
    [18]
])

walk_vs_stop = np.array([18])

for iteration in range(40):
    error_for_all_lights = 0
    for row_index in range(len(walk_vs_stop)):
        input = streetlights[row_index]
        goal_prediction = walk_vs_stop[row_index]

        prediction = input.dot(weights)
        error = (prediction - goal_prediction) ** 2
        error_for_all_lights += error

        delta = prediction - goal_prediction
        weights = weights - (alpha * (input * delta))

        print("Prediction:" + str(prediction))
    print("Error:" + str(error_for_all_lights) + "\\n")
```

# **Neural Networks Learn Correlation &amp; Up and Down Pressure**

Jaringan saraf pada dasarnya mempelajari hubungan korelasi antara masukan dan target. Selama pelatihan, setiap sampel data memberikan **upward pressure** (mendorong bobot naik) atau **downward pressure** (mendorong bobot turun) pada bobot-bobot jaringan.

# **Edge Cases: Overfitting &amp; Conflicting Pressure**

* **Overfitting**: Jaringan terlalu menghafal detail khusus data latih beserta *noise*\-nya sehingga gagal melakukan generalisasi pada data baru.
* **Conflicting Pressure**: Terjadi saat dua sampel data memberikan dorongan berlawanan pada bobot yang sama, sehingga jaringan harus mencari nilai kompromi terbaik.

# **Learning Indirect Correlation &amp; Stacking Neural Networks**

Jika data masukan dan target output tidak memiliki korelasi langsung, jaringan dua-layer (*single layer of weights*) akan gagal memprediksi target. Solusinya adalah menumpuk (*stacking*) beberapa jaringan dengan menambahkan **hidden layer** di tengah untuk menciptakan korelasi perantara (*intermediate correlation*).

# **Backpropagation: Long-Distance Error Attribution &amp; Why Does It Work?**

**Backpropagation** adalah metode *error attribution* jarak jauh. Algoritma ini menghitung seberapa besar setiap nodus pada *hidden layer* bertanggung jawab terhadap total eror di *output layer*. Caranya adalah membalikkan sinyal eror (*output delta*) melewati matriks bobot `weights_1_2` untuk menghasilkan `layer_1_delta` sebagai rata-rata tertimbang eror (`weighted average delta`).

# **Linear vs. Nonlinear &amp; The Secret to Sometimes Correlation**

Menumpuk matriks bobot linier tanpa fungsi aktivasi non-linier secara matematis hanya akan runtuh (*collapse*) menjadi 1 matriks linier tunggal. Fungsi aktivasi non-linier seperti **ReLU (Rectified Linear Unit)** (`relu(x) = max(0, x)`) mematikan nodus bernilai negatif menjadi 0. Sifat ini memberikan kemampuan **sometimes correlation** (korelasi kondisional) yang memungkinkan jaringan saraf dalam memecahkan masalah non-linier kompleks.

# **Your First Deep Neural Network &amp; Backpropagation in Code**

Menggabungkan seluruh komponen menjadi jaringan saraf 3-layer:
1. *Forward Prop*: `layer_1 = relu(np.dot(layer_0, weights_0_1))`, `layer_2 = np.dot(layer_1, weights_1_2)`.
2. *Calculate Output Delta*: `layer_2_delta = layer_2 - target`.
3. *Backpropagate Delta*: `layer_1_delta = layer_2_delta.dot(weights_1_2.T) * relu2deriv(layer_1)`. Turunan `relu2deriv` bernilai 1 jika nodus positif dan 0 jika negatif.
4. *Update Weights*: Memperbarui `weights_1_2` dan `weights_0_1` menggunakan perkalian luar masukan dan delta terkait.

```python
import numpy as np

np.random.seed(1)

def relu(x):
    return (x > 0) * x

alpha = 0.2
hidden_size = 4

streetlights = np.array([
    [18]
])

walk_vs_stop = np.array([[18]]).T

weights_0_1 = 2 * np.random.random((3, hidden_size)) - 1
weights_1_2 = 2 * np.random.random((hidden_size, 1)) - 1

layer_0 = streetlights
layer_1 = relu(np.dot(layer_0, weights_0_1))
layer_2 = np.dot(layer_1, weights_1_2)
```

```python
import numpy as np

np.random.seed(1)

def relu(x):
    return (x > 0) * x

def relu2deriv(output):
    return output > 0

streetlights = np.array([
    [18]
])

walk_vs_stop = np.array([[18]]).T

alpha = 0.2
hidden_size = 4

weights_0_1 = 2 * np.random.random((3, hidden_size)) - 1
weights_1_2 = 2 * np.random.random((hidden_size, 1)) - 1

for iteration in range(60):
    layer_2_error = 0
    for i in range(len(streetlights)):
        layer_0 = streetlights[i:i+1]
        layer_1 = relu(np.dot(layer_0, weights_0_1))
        layer_2 = np.dot(layer_1, weights_1_2)

        layer_2_error += np.sum((layer_2 - walk_vs_stop[i:i+1]) ** 2)

        layer_2_delta = (layer_2 - walk_vs_stop[i:i+1])
        layer_1_delta = layer_2_delta.dot(weights_1_2.T) * relu2deriv(layer_1)

        weights_1_2 -= alpha * layer_1.T.dot(layer_2_delta)
        weights_0_1 -= alpha * layer_0.T.dot(layer_1_delta)

    if (iteration % 10 == 9):
        print("Error:" + str(layer_2_error))
```
