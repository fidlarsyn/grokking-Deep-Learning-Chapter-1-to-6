# **Step 1: Predict**

Langkah pertama dalam pengoperasian jaringan saraf tiruan adalah membuat prediksi. Jaringan menerima data masukan (*input data*), memprosesnya melalui kombinasi bobot (*weights*), dan menghasilkan keluaran berupa prediksi.

# **A Simple Neural Network Making a Prediction**

Jaringan saraf tiruan paling sederhana terdiri dari 1 masukan, 1 bobot (*volume knob*), dan 1 keluaran. Operasi matematikanya adalah perkalian sederhana antara variabel *input* dan *weight*. Bobot berfungsi untuk memperbesar atau memperkecil dampak variabel masukan terhadap hasil prediksi.

```python
weight = 0.1

def neural\_network(input, weight):
  prediction = input \* weight
  return prediction

number\_of\_toes = [8.5, 9.5, 10, 9]
input = number\_of\_toes
pred = neural\_network(input, weight)
print(pred)
```

# **What is a Neural Network?**

Secara konseptual, jaringan saraf tiruan adalah sekumpulan bobot (*weights*) yang mengalikan nilai masukan untuk menghasilkan estimasi. Bobot tersebut menyimpan "pengetahuan" atau memori jaringan tentang hubungan antara input dan output.

# **What Does This Neural Network Do?**

Jaringan skala tunggal mengukur tingkat sensitivitas masukan terhadap keluaran. Misalnya, memberikan skor probabilitas kemenangan tim bisbol berdasarkan rata-rata jumlah jari kaki pemain.

# **Making a Prediction with Multiple Inputs**

Untuk membuat prediksi yang akurat, jaringan harus mampu mengombinasikan beberapa variabel masukan sekaligus (misalnya: *toes*, *win/loss record*, dan *number of fans*). Penggabungan ini dilakukan melalui operasi **weighted sum** (penjumlahan berbobot) antara perkalian tiap elemen masukan dengan bobotnya masing-masing.

```python
def w\_sum(a, b):
  assert(len(a) == len(b))
  output = 0
  for i in range(len(a)):
    output += (a[i] \* b[i])
  return output

weights = [0.1, 0.2, 0]

def neural\_network(input, weights):
  pred = w\_sum(input, weights)
  return pred

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input = [toes, wlrec, nfans]
pred = neural\_network(input, weights)
print(pred)
```

# **Multiple Inputs: What Does This Neural Network Do?**

*Weighted sum* menghitung tingkat kemiripan (*similarity*) antara vektor masukan dan vektor bobot. Bobot bernilai 0 akan mengabaikan masukan terkait, sedangkan bobot bernilai tinggi atau negatif akan memperbesar atau mengurangi skor prediksi secara signifikan.

# **Multiple Inputs: Complete Runnable Code**

Implementasi *weighted sum* dapat dilakukan menggunakan perulangan `for` bawaan Python atau secara efisien menggunakan fungsi perkalian titik (**dot product**) pada *library* **NumPy** (`input.dot(weights)`).

```python
import numpy as np

weights = np.array([0.1, 0.2, 0])

def neural\_network(input, weights):
  pred = input.dot(weights)
  return pred

toes = np.array([8.5, 9.5, 9.9, 9.0])
wlrec = np.array([0.65, 0.8, 0.8, 0.9])
nfans = np.array([1.2, 1.3, 0.5, 1.0])

input = np.array([toes, wlrec, nfans])
pred = neural\_network(input, weights)
print(pred)
```

# **Predicting with Multiple Inputs and Outputs**

Kombinasi dari kedua arsitektur sebelumnya menghasilkan jaringan dengan banyak input dan banyak output. Hubungan antar-layer diwakili oleh matriks bobot. Setiap nodus keluaran melakukan *weighted sum* independen terhadap seluruh vektor masukan.

# **Multiple Inputs and Outputs: How Does It Work?**

Arsitektur ini menjalankan perkalian vektor-matriks (*vector-matrix multiplication*). Operasi ini sama dengan melakukan beberapa *dot product* terpisah untuk setiap baris pada matriks bobot.

```python
weights = [
    [0.1, 0.1, -0.3], # hurt?
    [0.1, 0.2, 0.0], # win?
    [0.0, 1.3, 0.1] # sad?
]

def w\_sum(a, b):
  assert(len(a) == len(b))
  output = 0
  for i in range(len(a)):
    output += (a[i] \* b[i])
  return output

def vect\_mat\_mul(vect, matrix):
  assert(len(vect) == len(matrix))
  output =
  for i in range(len(vect)):
    output[i] = w\_sum(vect, matrix[i])
  return output

def neural\_network(input, weights):
  pred = vect\_mat\_mul(input, weights)
  return pred

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input = [toes, wlrec, nfans]
pred = neural\_network(input, weights)
print(pred)
```

# **Predicting on Predictions (Stacking Neural Networks)**

Jaringan saraf dalam (*deep neural networks*) dibuat dengan menumpuk beberapa jaringan secara berurutan. Output dari layer pertama (*hidden layer*) dialirkan sebagai input untuk layer berikutnya.

```python
import numpy as np

# toes % win # fans
ih\_wgt = np.array([ [0.1, 0.2, -0.1], # hid
                     [-0.1, 0.1, 0.9], # hid[23]
                     [0.1, 0.4, 0.1] # hid[1]
]).T

# hid hid[23] hid[1]
hp\_wgt = np.array([ [0.3, 1.1, -0.3], # hurt?
                     [0.1, 0.2, 0.0], # win?
                     [0.0, 1.3, 0.1] # sad?
]).T

weights = [ih\_wgt, hp\_wgt]

def neural\_network(input, weights):
  hid = input.dot(weights)
  pred = hid.dot(weights[23])
  return pred

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input = [toes, wlrec, nfans]
pred = neural\_network(input, weights)
print(pred)
```
