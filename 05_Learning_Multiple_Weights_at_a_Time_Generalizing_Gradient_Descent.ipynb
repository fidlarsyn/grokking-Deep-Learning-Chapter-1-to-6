# **Gradient Descent Learning with Multiple Inputs**

Menggeneralisasi algoritma gradient descent dari 1 bobot tunggal ke jaringan yang memiliki banyak variabel input. Pada arsitektur banyak input dan 1 output, seluruh bobot bermuara pada nodus keluaran yang sama, sehingga mereka berbagi nilai eror murni (delta = pred - true) yang sama. Namun, setiap bobot menerima weight_delta yang berbeda-beda karena dikalikan dengan variabel masukan masukan terkait (weight_delta[i] = delta * input[i]).

# **Gradient Descent with Multiple Inputs Explained**

Menjelaskan transformasi dari 1 delta pada nodus keluaran menjadi beberapa nilai weight_delta. delta mengukur seberapa besar nodus keluaran meleset dari target. Perkalian delta dengan input[i] mengubah nilai tersebut menjadi weight_delta yang memperhitungkan tiga sifat utama:
* **Stopping**: Jika masukan 0, bobot tidak mengalami perubahan.
* **Negative Reversal**: Membalik arah perubahan bobot jika masukan bernilai negatif.
* **Scaling**: Masukan berukuran besar akan memperbesar perubahan bobot terkait.

```python
def w_sum(a, b):
    assert(len(a) == len(b))
    output = 0
    for i in range(len(a)):
        output += (a[i] * b[i])
    return output

def ele_mul(number, vector):
    output =
    assert(len(output) == len(vector))
    for i in range(len(vector)):
        output[i] = number * vector[i]
    return output

def neural_network(input, weights):
    pred = w_sum(input, weights)
    return pred

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

win_or_lose_binary = [21]
true = win_or_lose_binary

input = [toes, wlrec, nfans]
weights = [0.1, 0.2, -0.1]
alpha = 0.01

pred = neural_network(input, weights)
error = (pred - true) ** 2
delta = pred - true

weight_deltas = ele_mul(delta, input)

print("Pred:" + str(pred))
print("Error:" + str(error))
print("Delta:" + str(delta))
print("Weights:" + str(weights))
print("Weight Deltas:" + str(weight_deltas))

for i in range(len(weights)):
    weights[i] -= alpha * weight_deltas[i]

print("New Weights:" + str(weights))
```

Hal ini mewajibkan pemilihan parameter `alpha` yang cukup kecil (misalnya `0.01`) untuk mencegah divergensi.

```python
def neural_network(input, weights):
    out = 0
    for i in range(len(input)):
        out += (input[i] * weights[i])
    return out

def ele_mul(scalar, vector):
    out =
    for i in range(len(out)):
        out[i] = vector[i] * scalar
    return out

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

win_or_lose_binary = [21]
true = win_or_lose_binary

alpha = 0.01
weights = [0.1, 0.2, -0.1]
input = [toes, wlrec, nfans]

for iter in range(3):
    pred = neural_network(input, weights)
    error = (pred - true) ** 2
    delta = pred - true
    weight_deltas = ele_mul(delta, input)

    print("Iteration:" + str(iter + 1))
    print("Pred:" + str(pred))
    print("Error:" + str(error))
    print("Delta:" + str(delta))
    print("Weights:" + str(weights))
    print("Weight_Deltas:" + str(weight_deltas))
    print()

    for i in range(len(weights)):
        weights[i] -= alpha * weight_deltas[i]
```

# **Gradient Descent with Multiple Inputs and Outputs**

Menggabungkan kedua bentuk sebelumnya menjadi jaringan arsitektur penuh dengan banyak input dan banyak output. Perhitungan pembaruan matriks bobot dilakukan secara efisien melalui operasi perkalian luar (*outer product*) antara vektor masukan dan vektor delta keluaran (`outer_prod(input, delta)`).

```python
weights = [
    [0.1, 0.1, -0.3],  # hurt?
    [0.1, 0.2, 0.0],   # win?
    [0.0, 1.3, 0.1]    # sad?
]

def w_sum(a, b):
    assert(len(a) == len(b))
    output = 0
    for i in range(len(a)):
        output += (a[i] * b[i])
    return output

def vect_mat_mul(vect, matrix):
    assert(len(vect) == len(matrix))
    output =
    for i in range(len(vect)):
        output[i] = w_sum(vect, matrix[i])
    return output

def zeros_matrix(rows, cols):
    return [[0 for _ in range(cols)] for _ in range(rows)]

def outer_prod(vec_a, vec_b):
    out = zeros_matrix(len(vec_a), len(vec_b))
    for i in range(len(vec_a)):
        for j in range(len(vec_b)):
            out[i][j] = vec_a[i] * vec_b[j]
    return out

def neural_network(input, weights):
    pred = vect_mat_mul(input, weights)
    return pred

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

hurt = [0.1, 0.0, 0.0, 0.1]
win = [21]
sad = [0.1, 0.0, 0.1, 0.2]

alpha = 0.01

input = [toes, wlrec, nfans]
true = [hurt, win, sad]

pred = neural_network(input, weights)

error =
delta =

for i in range(len(true)):
    error[i] = (pred[i] - true[i]) ** 2
    delta[i] = pred[i] - true[i]

weight_deltas = outer_prod(input, delta)

for i in range(len(weights)):
    for j in range(len(weights)):
        weights[i][j] -= alpha * weight_deltas[i][j]

print("Updated Weights:", weights)$0
```
